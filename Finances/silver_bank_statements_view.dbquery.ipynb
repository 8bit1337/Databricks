```sql
create or replace view finances.02_silver.bank_statements as 
  select
    to_date(`Date Posted`, 'yyyyMMdd') as Date,
    `Transaction Type` as Category,
    right(left(raw.Description, 3), 2) as Code,
    t.Name as Type,
    regexp_replace(
      trim(right(raw.Description, length(raw.Description) - 4)),
      ' +',
      ' '
    ) as Description,
    ` Transaction Amount` as Amount
  from finances.`01_bronze`.bank_statements raw
  left join finances.refdata.code__type t on t.Code = right(left(raw.Description, 3), 2)
```
