```sql
-- create or replace view finances.02_silver.creditcard_statements as 
    select
        Account,
        to_date(`Posted Date`, 'yyyyMMdd') as Date,
        regexp_replace(Payee, ' +', ' ') as Description,
        case when Amount < 0 then 'Credit' else 'Debit' end as EntryIs,
        case when Amount < 0 then -Amount else Amount end as Amount
    from finances.01_bronze.cc_amazon_statements am
    union all
    select
        Account,
        to_date(`Posting Date`, 'yyyyMMdd') as Date,
        regexp_replace(mc.Description, ' +', ' ') as Description,
        case when `Transaction Amount` < 0 then 'Credit' else 'Debit' end as EntryIs,
        case when `Transaction Amount` < 0 then -`Transaction Amount` else `Transaction Amount` end as Amount
    from finances.01_bronze.cc_bmo_statements mc
    union all
    select
        Account,
        to_date(`Posted Date`, 'yyyyMMdd') as Date,
        regexp_replace(rg.`Merchant Name`, ' +', ' ') as Description,
        case when cast(replace(Amount,'$','') as double) < 0 then 'Credit' else 'Debit' end as EntryIs,
        case when cast(replace(Amount,'$','') as double) < 0 then -cast(replace(Amount,'$','') as double) else cast(replace(Amount,'$','') as double) end as Amount
    from finances.01_bronze.cc_rogers_statements rg
```
