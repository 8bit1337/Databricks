```sql
create or replace view finances.02_silver.creditcard_statements as 
    select
        'BMO CC' as Card,
        to_date(`Posting Date`, 'yyyyMMdd') as Date,
        regexp_replace(mc.Description, ' +', ' ') as Description,
        `Transaction Amount` as Amount
    from finances.01_bronze.bmo_cc_statements mc
    union all
    select
        'Amazon CC' as Card,
        to_date(`Posted Date`, 'yyyyMMdd') as Date,
        regexp_replace(Payee, ' +', ' ') as Description,
        Amount
    from finances.01_bronze.amazon_cc_statements am
```
