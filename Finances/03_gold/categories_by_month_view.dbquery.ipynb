```sql
create or replace view finances.03_gold.categories_by_month as
	with actuals as (
		select 
			cast(date_trunc('month', Date) as date) as Month,
			BudgetCategory,
			EntryIs,
			round(sum(Amount),2) as Amount
		from finances.03_gold.all_transactions 
		group by Month, BudgetCategory, EntryIs
	)

	select
		
		'Actual' as MetricName,
		a.BudgetCategory,
		a.Month,
		a.EntryIs,
		a.Amount as MetricValue
	from actuals a
	union all
	select
		'Budget' as MetricName,
		b.Category as BudgetCategory,
		t.Month,
		null as EntryIs,		
		b.Budget as MetricValue
	from finances.refdata.budget b
	join (select distinct Month from actuals) t
```
